```python
# Precision

#Notas: Para a precisión se obtine un vector de relevancia binaria de longitud n, donde n es el número de documentos recuperados.
#Cada elemento del vector es 1 si el documento es relevante y 0 si no lo es. La precisión se calcula como la suma de los elementos del vector de relevancia binaria dividido por n.

def precision(relevant_documents, retrieved_documents):
    """
    Calcula la precisión de un sistema de recuperación de información.
    
    Parámetros:
    relevant_documents (list): Lista de documentos relevantes.
    retrieved_documents (list): Lista de documentos recuperados por el sistema.
    
    Retorna:
    float: Precisión del sistema.
    """
    # Convertimos las listas a conjuntos para facilitar la intersección
    relevant_set = set(relevant_documents)
    retrieved_set = set(retrieved_documents)
    
    # Calculamos la intersección de los conjuntos
    true_positives = len(relevant_set.intersection(retrieved_set))
    
    # Calculamos la precisión
    if len(retrieved_set) == 0:
        return 0.0  # Evitamos división por cero
    precision_value = true_positives / len(retrieved_set)
    
    return precision_value
```
